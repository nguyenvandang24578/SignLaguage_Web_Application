### Unsloth Congfiguration


````python
from unsloth import FastLanguageModel
import torch

# Config
model_name = "unsloth/Qwen3-1.7B-unsloth-bnb-4bit"
repo_name  = "VyDat/Qwen3_translate_v2"
data       = "VyDat/VSL-Dataset-10K"
max_seq_length = 128
SYS_PROMPT = "Bạn là một chuyên gia ngôn ngữ, có thể tái cấu trúc câu từ ngôn ngữ nói Tiếng Việt thành cấu trúc câu Ngôn ngữ ký hiệu Việt Nam, hãy chuyển đổi câu sau:"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = model_name,
    dtype = torch.float16,
    max_seq_length = max_seq_length,
    load_in_4bit = True,
    load_in_8bit = False,
    full_finetuning = False,
)
````

````python
model = FastLanguageModel.get_peft_model(
    model,
    r = 64,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj",],
    lora_alpha = 128,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = "unsloth",
    random_state = 42,
    use_rslora = False,
    loftq_config = None,
)
````

### Data Prep


````python
from unsloth.chat_templates import get_chat_template
tokenizer = get_chat_template(
    tokenizer,
    chat_template = "qwen3-instruct",
)
````

````python
from datasets import load_dataset
dataset = load_dataset(data, split = "train")
print(f"Dataset size: {len(dataset)}")
print(f"Columns: {dataset.column_names}")
print(f"Sample: {dataset[0]}")
````

```
{"conversations": [
  {"from": "system", "value": SYS_PROMPT},
  {"from": "human",  "value": source},
  {"from": "gpt",    "value": target}
]}
```


````python
def to_conversations(example):
    return {
        "conversations": [
            {"from": "system", "value": SYS_PROMPT},
            {"from": "human",  "value": example["source"]},
            {"from": "gpt",    "value": example["target"]},
        ]
    }

dataset = dataset.map(to_conversations, remove_columns=dataset.column_names)
````

````python
from unsloth.chat_templates import standardize_data_formats
dataset = standardize_data_formats(dataset)
````

````python
dataset[0]
````

````python
def formatting_prompts_func(examples):
    convos = examples["conversations"]
    texts = [tokenizer.apply_chat_template(convo, tokenize=False, add_generation_prompt=False) for convo in convos]
    return {"text": texts}

dataset = dataset.map(formatting_prompts_func, batched=True)
````

````python
print(dataset[0]["text"])
````

````python
data_split = dataset.train_test_split(test_size=0.1, shuffle=True, seed=42)
data_train = data_split["train"]
data_eval  = data_split["test"]
print(f"Train: {len(data_train)}, Eval: {len(data_eval)}")
````

<a name="Train"></a>

### Train the model

Use `SFTTrainer` with `SFTConfig`


````python
from trl import SFTTrainer, SFTConfig

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = data_train,
    eval_dataset = data_eval,
    args = SFTConfig(
        dataset_text_field = "text",
        per_device_train_batch_size = 8,
        gradient_accumulation_steps = 4,
        warmup_steps = 100,
        num_train_epochs = 2,
        learning_rate = 3e-4,
        fp16 = True, #not torch.cuda.is_bf16_supported(False),
        bf16 = False, #torch.cuda.is_bf16_supported(),
        logging_steps = 25,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 42,
        output_dir = "outputs",
        save_strategy = "steps",
        save_steps = 100,
        eval_strategy = "steps",
        eval_steps = 200,
        save_total_limit = 3,
        report_to = "wandb",
        max_seq_length = max_seq_length,
    ),
)
````

We use `train_on_responses_only` to only train on the assistant outputs and ignore the instruction tokens.


````python
from unsloth.chat_templates import train_on_responses_only
trainer = train_on_responses_only(
    trainer,
    instruction_part = "<|im_start|>user\n",
    response_part    = "<|im_start|>assistant\n",
)
````

## Evaluate Metrics


````python
import torch
import sacrebleu
from tqdm import tqdm
from datasets import load_dataset

# Load lại dataset GỐC có cột source/target
raw_dataset = load_dataset(data, split="train")
raw_split   = raw_dataset.train_test_split(test_size=0.1, shuffle=True, seed=42)
data_eval_raw = raw_split["test"]

def evaluate_translation(model, tokenizer, eval_dataset, max_new_tokens=128, batch_size=16):
    model.eval()

    all_preds = []
    all_refs  = []

    for i in tqdm(range(0, len(eval_dataset), batch_size), desc="Evaluating"):
        batch = eval_dataset[i : i + batch_size]

        prompts = []
        for src in batch["source"]:
            messages = [
                {"role": "system", "content": SYS_PROMPT},
                {"role": "user",   "content": src},
            ]
            prompts.append(
                tokenizer.apply_chat_template(
                    messages,
                    tokenize=False,
                    add_generation_prompt=True,
                )
            )

        inputs = tokenizer(
            prompts,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=max_seq_length,
        ).to("cuda")

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

        input_len = inputs["input_ids"].shape[1]
        generated = outputs[:, input_len:]

        preds = tokenizer.batch_decode(generated, skip_special_tokens=True)
        refs  = batch["target"]

        all_preds.extend([p.strip() for p in preds])
        all_refs.extend([r.strip() for r in refs])

    bleu_result = sacrebleu.corpus_bleu(all_preds, [all_refs])

    print(f"\n--- Kết quả đánh giá ---")
    print(f"BLEU Total: {bleu_result.score:.2f}")
    print(f"BLEU-1    : {bleu_result.precisions[0]:.2f}")
    print(f"BLEU-2    : {bleu_result.precisions[1]:.2f}")
    print(f"BLEU-3    : {bleu_result.precisions[2]:.2f}")
    print(f"BLEU-4    : {bleu_result.precisions[3]:.2f}")

    print(f"\nSamples:")
    for i in range(min(5, len(all_preds))):
        print(f"  Source : {eval_dataset[i]['source']}")
        print(f"  Target : {all_refs[i]}")
        print(f"  Predict: {all_preds[i]}")
        print()

    return {
        "bleu_total": round(bleu_result.score, 2),
        "bleu_1":     round(bleu_result.precisions[0], 2),
        "bleu_2":     round(bleu_result.precisions[1], 2),
        "bleu_3":     round(bleu_result.precisions[2], 2),
        "bleu_4":     round(bleu_result.precisions[3], 2),
    }

# Gọi đúng dataset gốc
results = evaluate_translation(model, tokenizer, data_eval_raw)
````
